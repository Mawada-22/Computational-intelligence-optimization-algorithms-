```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def fitness_function(x1, x2):
    return np.sin(2*x1 - 1.57) + 3*np.cos(x2) + 0.5*x1
```

```python
def particle_swarm_optimization(num_iterations=500):
    # Parameters
    num_particles = 50
    num_dimensions = 2
    v_max = [-0.1, 1]

    # Initialize population
    population_positions = np.random.uniform(-2, 3, size=(num_particles, num_dimensions))
    population_velocities = np.random.uniform(-0.1, 1, size=(num_particles, num_dimensions))

    # Initialize personal best and global best
    pbest_positions = population_positions.copy()
    pbest_fitness = fitness_function(population_positions[:,0], population_positions[:,1])
    gbest_index = np.argmax(pbest_fitness)
    gbest_fitness = pbest_fitness[gbest_index]
    gbest_position = population_positions[gbest_index]

    # Constants
    c1 = c2 = 2

    # Main loop
    for i in range(num_iterations):
        # Evaluate fitness
        fitness_values = fitness_function(population_positions[:,0], population_positions[:,1])

        # Update personal best
        update_indices = fitness_values > pbest_fitness
        pbest_positions[update_indices] = population_positions[update_indices]
        pbest_fitness[update_indices] = fitness_values[update_indices]

       
        gbest_index = np.argmax(fitness_values)
        if fitness_values[gbest_index] > gbest_fitness:
            gbest_fitness = fitness_values[gbest_index]
            gbest_position = population_positions[gbest_index]

        
        r1 = np.random.rand(num_particles, num_dimensions)
        r2 = np.random.rand(num_particles, num_dimensions)
        population_velocities = population_velocities + c1 * r1 * (pbest_positions - population_positions) + c2 * r2 * (gbest_position - population_positions)
        population_velocities = np.clip(population_velocities, -0.1, 1)
        population_positions = population_positions + population_velocities
        population_positions = np.clip(population_positions, -2, 3)

    return gbest_position

```

```python
# Define num_particles and num_dimensions
num_particles = 50
num_dimensions = 2

# Run PSO and get the global best position
final_gbest_position = particle_swarm_optimization()
print("Global Best Position:", final_gbest_position)

# Plot the movement of particles
population_positions = np.random.uniform(-2, 3, size=(num_particles, num_dimensions))  # Redefine here
plt.figure(figsize=(8, 6))
plt.plot(population_positions[:,0], population_positions[:,1], 'o', label='Particles')
plt.plot(final_gbest_position[0], final_gbest_position[1], 'ro', label='Global Best')
plt.xlabel('x1')
plt.ylabel('x2')
plt.title('Particle Swarm Optimization')
plt.legend()
plt.grid(True)
plt.show()

plt.show()

```
